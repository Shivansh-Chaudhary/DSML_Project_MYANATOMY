```python
#import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load the dataset
df=pd.read_csv('/content/salary_dataset (1) (1).csv')
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe()
```

```python
#checking for missing values
df.isnull().sum()
```

```python
#Experience Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['Experience'],bins=10,kde=True)
plt.title('Experience Distribution')
plt.xlabel('Years of Experience')
plt.ylabel('Number of Employees')
plt.show
```

```python
#Number of employees by job role
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='JobRole')
plt.title('Number of Employees by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.show()
```

```python
#salary by jobrole
#bi-variate analysis(including two Variables)
plt.figure(figsize=(10,5))
sns.barplot(data=df,x='JobRole',y='Salary')
plt.title('Salary by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Average Salary')
plt.xticks(rotation=45)
plt.show()
```

```python
#Salary by Education
plt.figure(figsize=(8,5))
sns.barplot(data=df,x='Education',y='Salary')
plt.title('Average Salary by Educaation')
plt.xlabel('Education')
plt.ylabel('Average Salary')
plt.show()
```

```python
#salary by location
plt.figure(figsize=(10,5))
sns.barplot(data=df,x='Location',y='Salary')
plt.title('Average Salary by Location')
plt.xlabel('Location')
plt.ylabel('Average Salary')
plt.show()
```

```python
#Experience Vs Salary
plt.figure(figsize=(10,5))
sns.scatterplot(data=df,x='Experience',y='Salary')
plt.title('Experience Vs Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
#with regression line
plt.figure(figsize=(10,5))
sns.regplot(data=df,x='Experience',y='Salary')
plt.title('Experience Vs Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
#salary distribution with education
plt.figure(figsize=(10,5))
sns.boxplot(data=df,x='Education',y='Salary')
plt.title('Salary Distribution by Education')
plt.xlabel('Education')
plt.ylabel('Salary')
plt.show()
```
